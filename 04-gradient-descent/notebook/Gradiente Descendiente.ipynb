Lo primero que debemos hacer es importar las librerias necesarias para relizar la implementacion y verificar que esten instaladas correctamente

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from ipywidgets import interact, FloatSlider

print("Todas las librerias necesarias estan instaladas correctamente")

# Auto-added to ensure figures render
try:
    plt.tight_layout()
except Exception:
    pass
plt.show()
```

Ahora debemos cargar el archivo de excel y para extraer las variables utiles

```python


data = pd.read_excel("Real_estate_valuation_data_set.xlsx")

X = data.iloc[:, 1:-1].values
y = data.iloc[:, -1].values.reshape(-1, 1)

data.head(5)
```

```python

scaler = StandardScaler()
X = scaler.fit_transform(X)
```

Con esto debemos calcular la función de costo

```python

def compute_cost(X_b, y, theta):
    m = len(y)
    return (1/(2*m)) * np.sum((X_b.dot(theta) - y) ** 2)
```

Y luego la función de gradiente descendiente

```python

def gradient_descent(X, y, alpha, iterations):
    m, n = X.shape
    X_b = np.c_[np.ones((m, 1)), X]  # Agregar columna de 1’s (bias)
    theta = np.zeros((n+1, 1))
    cost_history = []

    for _ in range(iterations):
        gradients = (1/m) * X_b.T.dot(X_b.dot(theta) - y)
        theta -= alpha * gradients
        cost = compute_cost(X_b, y, theta)
        cost_history.append(cost)
    return theta, cost_history
```

Por ultimo, se plotea la grafica interactiva

```python

def run_gradient_descent(alpha):
    theta, cost_history = gradient_descent(X, y, alpha, iterations=100)
    
    plt.figure(figsize=(9,7))
    plt.plot(range(len(cost_history)), cost_history, 'b-')
    plt.xlabel("Numero de iteraciones")
    plt.ylabel("Costo")
    plt.title(f"Función de costo interactiva con: α={alpha:.4f}")
    plt.show()
    
    print("Parámetros finales (theta):")
    print(theta.flatten())
    
interact(run_gradient_descent,
         alpha=FloatSlider(min=0.0000001, max=0.5, step=0.001, value=0.01))
```
